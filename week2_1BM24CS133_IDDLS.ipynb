## 8-Puzzle Solver using Iterative Deepening Depth-Limited Search (IDDLS)

This notebook demonstrates the implementation of the Iterative Deepening Depth-Limited Search (IDDLS) algorithm to solve the classic 8-puzzle game.

### Game Representation

The 8-puzzle game is represented as a 3x3 grid. The state of the puzzle is a tuple of tuples, making it immutable and hashable, which is useful for checking visited states. The blank space is represented by `0`.

```python
from collections import deque

# Define the goal state
GOAL_STATE = ((1, 2, 3), (4, 5, 6), (7, 8, 0))

def find_blank(state):
    """Finds the position (row, col) of the blank (0) tile."""
    for r in range(3):
        for c in range(3):
            if state[r][c] == 0:
                return r, c
    return None

def get_possible_moves(r, c):
    """Returns a list of possible (dr, dc) movements for a tile at (r, c)."""
    moves = []
    if r > 0: # Move up
        moves.append((-1, 0))
    if r < 2: # Move down
        moves.append((1, 0))
    if c > 0: # Move left
        moves.append((0, -1))
    if c < 2: # Move right
        moves.append((0, 1))
    return moves

def apply_move(state, blank_pos, move):
    """Applies a move to the state and returns the new state and its path."""
    r, c = blank_pos
    dr, dc = move

    new_r, new_c = r + dr, c + dc

    # Create a mutable list of lists from the current state tuple of tuples
    new_state_list = [list(row) for row in state]

    # Swap the blank tile with the tile at the new position
    new_state_list[r][c], new_state_list[new_r][new_c] = new_state_list[new_r][new_c], new_state_list[r][c]

    # Convert back to tuple of tuples for the new state
    return tuple(tuple(row) for row in new_state_list)

def print_state(state):
    """Prints the 8-puzzle state in a readable 3x3 format."""
    for row in state:
        print(" ".join(map(str, row)).replace('0', ' '))
    print()
```

```python
def dls(initial_state, limit):
    """Depth-Limited Search algorithm for the 8-puzzle."""
    # A stack to store (state, path) tuples
    # Path is a list of states from initial to current
    stack = [(initial_state, [initial_state])]
    visited = set()
    visited.add(initial_state)

    while stack:
        current_state, path = stack.pop()

        if current_state == GOAL_STATE:
            return path # Goal reached within limit

        # If current depth exceeds limit, prune this path
        if len(path) - 1 == limit:
            continue

        blank_r, blank_c = find_blank(current_state)

        # Get valid moves and explore neighbors
        for dr, dc in get_possible_moves(blank_r, blank_c):
            new_state = apply_move(current_state, (blank_r, blank_c), (dr, dc))
            if new_state not in visited:
                visited.add(new_state)
                stack.append((new_state, path + [new_state]))
    return None # Goal not found within limit
```

```python
def idds(initial_state, max_depth=20):
    """Iterative Deepening Depth-Limited Search algorithm for the 8-puzzle."""
    for depth_limit in range(max_depth + 1):
        print(f"Searching with depth limit: {depth_limit}")
        solution = dls(initial_state, depth_limit)
        if solution is not None:
            return solution
    return None # Goal not found within max_depth
```

```python
# Example initial state (solvable)
initial_state = (
    (1, 2, 3),
    (0, 4, 6),
    (7, 5, 8)
)

print("Initial State:")
print_state(initial_state)

print("Goal State:")
print_state(GOAL_STATE)

print("Solving the 8-puzzle using IDDLS...")
solution_path = idds(initial_state, max_depth=10) # Set a reasonable max_depth

if solution_path:
    print(f"\nSolution found in {len(solution_path) - 1} steps!")
    for i, state in enumerate(solution_path):
        print(f"Step {i}:")
        print_state(state)
else:
    print("\nNo solution found within the maximum depth limit.")
```
