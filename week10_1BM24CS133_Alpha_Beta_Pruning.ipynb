```python
import math

# Global dictionaries to store the user-defined game tree
tree = {}       # Maps state -> list of (action, next_state)
utilities = {}  # Maps terminal_state -> numeric utility value


def terminal_test(state):
    return state in utilities


def utility(state):
    return utilities[state]


def actions(state):
    return [action for action, _ in tree.get(state, [])]


def result(state, action):
    for act, next_state in tree.get(state, []):
        if act == action:
            return next_state
    return None


def max_value(state, alpha, beta):
    if terminal_test(state):
        return utility(state)

    v = -math.inf
    for a in actions(state):
        next_state = result(state, a)
        v = max(v, min_value(next_state, alpha, beta))
        if v >= beta:
            print(f"[PRUNED at MAX node '{state}'] v ({v}) >= beta ({beta})")
            return v
        alpha = max(alpha, v)
    return v


def min_value(state, alpha, beta):
    if terminal_test(state):
        return utility(state)

    v = math.inf
    for a in actions(state):
        next_state = result(state, a)
        v = min(v, max_value(next_state, alpha, beta))
        if v <= alpha:
            print(f"[PRUNED at MIN node '{state}'] v ({v}) <= alpha ({alpha})")
            return v
        beta = min(beta, v)
    return v


def alpha_beta_search(state):
    best_value = -math.inf
    best_action = None
    alpha = -math.inf
    beta = math.inf

    for a in actions(state):
        next_state = result(state, a)
        v = min_value(next_state, alpha, beta)
        if v > best_value:
            best_value = v
            best_action = a
        alpha = max(alpha, best_value)

    return best_action, best_value


def build_tree_from_input():
    root = input("Enter the root node name (e.g., A): ").strip()

    num_internal = int(input("Enter number of non-terminal (internal) nodes: "))
    print("\nFor each internal node, enter its child nodes separated by spaces.")
    for _ in range(num_internal):
        node = input("  Internal node name: ").strip()
        children = input(f"  Children of '{node}' (left to right): ").strip().split()
        # Action is labeled as "node->child"
        tree[node] = [(f"{node}->{child}", child) for child in children]

    num_terminals = int(input("\nEnter number of terminal (leaf) nodes: "))
    print("For each terminal node, enter its name and utility value separated by a space:")
    for _ in range(num_terminals):
        entry = input("  Leaf name and value (e.g., L1 5): ").strip().split()
        leaf_name, val = entry[0], float(entry[1])
        utilities[leaf_name] = int(val) if val.is_integer() else val

    return root





if __name__ == "__main__":
    root_node = build_tree_from_input()
    print("\n--- Running Alpha-Beta Search ---")
    action, value = alpha_beta_search(root_node)
    print(f"\nValue of Root Node ({root_node}): {value}")
    print(f"Optimal First Action: {action}")
```
